# **Image Translation**

Write a Python program using OpenCV to perform image translation on the **"Given image"**.

The program should:

1. Import `cv2`, `numpy`, and `matplotlib.pyplot`.
2. Read the image **"given** in color mode.
3. Create a **translation matrix** that moves the image  
   **50 pixels to the right** and **80 pixels down**.
4. Apply the translation
5. Display the following using a **1 × 2 layout** in Matplotlib:  
   - **Original Image**  
   - **Translated Image**


```python
# Import 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("baseball.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python

rows, cols, channels = img.shape

M = np.float32([[1, 0, 50],
                [0, 1, 80]])

translated = cv2.warpAffine(img, M, (cols, rows))
```

```python
# Display the image using Matplotlib
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(translated)
plt.title("Translated Image")
plt.axis("off")

plt.show()
```

# **Image Shearing**

Write a Python program using OpenCV to perform image shearing on the **given image**.

The program should:

1. Create a **shearing transformation matrix** to apply:
   - **Horizontal shearing**  
   - **Vertical shearing**
2. Apply the shearing transformations.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **Horizontally Sheared Image**  
   - **Vertically Sheared Image**


```python
rows, cols, channels = img.shape

M_horizontal = np.float32([[1, 0.5, 0],
                           [0, 1, 0]])

M_vertical = np.float32([[1, 0, 0],
                         [0.5, 1, 0]])
```

```python
horizontal = cv2.warpAffine(img, M_horizontal, (cols, rows))
vertical = cv2.warpAffine(img, M_vertical, (cols, rows))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(horizontal)
plt.title("Horizontally Sheared Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(vertical)
plt.title("Vertically Sheared Image")
plt.axis("off")

plt.show()
```

# **Image Rotation**

Write a Python program using OpenCV to perform **image rotation** on the **given image**.

The program should:

1. Rotate the image by:  
   - **45 degrees**  
   - **90 degrees**
2. Use `cv2.getRotationMatrix2D` and `cv2.warpAffine`.
3. Display the following using a **1 × 3 layout** in Matplotlib:  
   - **Original Image**  
   - **45° Rotated Image**  
   - **90° Rotated Image**


```python
rows, cols, channels = img.shape

center = (cols // 2, rows // 2)

M45 = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated45 = cv2.warpAffine(img, M45, (cols, rows))

M90 = cv2.getRotationMatrix2D(center, 90, 1.0)
rotated90 = cv2.warpAffine(img, M90, (cols, rows))
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(rotated45)
plt.title("45 Degree Rotation")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(rotated90)
plt.title("90 Degree Rotation")
plt.axis("off")

plt.show()
```
